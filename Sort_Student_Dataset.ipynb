```python
import pandas as pd
```

```python
df = pd.read_csv("/content/student_data.csv")
```

```python
df.head()
```

```python
print("\nDataset Shape:", df.shape)
```

```python
ascending_sort = df.sort_values(
    by='age',
    ascending=True
)

display(ascending_sort.head(10))
```

```python
descending_sort = df.sort_values(
    by='age',
    ascending=False
)

display(descending_sort.head(10))
```

```python
school_sort = df.sort_values(
    by='school',
    ascending=True
)

display(school_sort.head(10))
```

```python
study_sort = df.sort_values(
    by='studytime',
    ascending=False
)

display(study_sort.head(10))
```

```python
absence_sort = df.sort_values(
    by='absences',
    ascending=True
)

display(absence_sort.head(10))
```

```python

grade_descending = df.sort_values(by='G3', ascending=False)

reset_sort = grade_descending.reset_index(drop=True)

display(reset_sort.head(10))
```
